# Data Cleaning

Loads the raw MXMH survey data, reports missing values, and saves the cleaned dataset used by
the analysis notebooks. Cleaning drops the unused columns (`mc.UNUSED_COLUMNS`) and then every
row with a missing value in the remaining columns.

```python
import mxmh_common as mc

raw = mc.load_raw()
print(f"The MXMH dataset has {raw.shape[0]} rows and {raw.shape[1]} columns")
```

## Columns

Column names, non-null counts and data types.

```python
raw.info(verbose=True)
```

## Missing Values

```python
missing_counts = raw.isna().sum()
missing_counts[missing_counts > 0].to_frame("Missing")
```

## Save the Cleaned Dataset

```python
clean = mc.save_clean_dataset()
print(f"Saved {mc.CLEAN_DATA_PATH.name}: {clean.shape[0]} rows and {clean.shape[1]} columns")
```
